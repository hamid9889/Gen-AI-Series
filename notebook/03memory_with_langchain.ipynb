```python
llm = ChatOpenAI(model ="gpt-4o" )
```

```python
history = []
while True:
    query = input("user: ")
    if query.lower() in ["exit", "bye", "quit"]:
        print("goodbye")
        break
    
    # Corrected dictionary keys
    history.append({"role": "user", "content": query})
    print("user:", query)
    
    res = llm.invoke(history)
    history.append({"role": "assistant", "content": res.content})
    print("AI:", res.content, "\n")
```
