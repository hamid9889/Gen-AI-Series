```python
from pydantic import BaseModel , Field
from typing import List
```

```python


class ResponseStructure(BaseModel):
    name:str = Field(description = "complete Name")
    email:str = Field(description = "Email Address")
    age:str = Field(description = "Age")

strectured_llm = llm.with_structured_output(ResponseStructure)
    
```

```python
res = strectured_llm.invoke(f" please give me only name email from this text : {text}")
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie title")
    year:str = Field(description="Movie releasesc year")
    

class AllMovies(BaseModel):
    movies:list[Movies]


movie_llm = llm.with_structured_output(AllMovies)   
 
    
```
