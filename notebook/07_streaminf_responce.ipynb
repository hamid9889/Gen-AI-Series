```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-4")
```

```python
question = " can ypu explain me GEN AI"
```

```python
res = llm.stream(question)
for chunk in res:
    print(chunk.content, end = "")
```

```python
res = llm.invoke(question)
```
