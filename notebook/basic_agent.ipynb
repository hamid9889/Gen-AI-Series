```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
       it will return the sum of two numbers
       
       Args:
       a : NUmber One
       b : Number Two
    
    
    """
    return a +b
 
@tool
def multiply_numbers(a:int, b:int):
    """
       it will return the product of two numbers
       
       Args:
       a : NUmber One
       b : Number Two
    
    
    """
    return a * b 
```

```python
add_numbers.invoke({"a":12, "b":8})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm = ChatOpenAI(model="gpt-4")
```

```python
agent = create_agent(
    model = llm, 
    tools=[add_numbers, multiply_numbers], 
    system_prompt="you are a math teacher , and always use tool for calculation "
    
    
)
```

```python
res = agent.invoke({"messages":[{"role":"user", "content":"what is 2+30 * 5 ?"}]})
```
