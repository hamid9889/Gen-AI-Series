```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """
My dear friends, students, and respected farmers,

Today, I want to speak from the heart about three pillars that define the future of our nation:
our youth, our education system, and our farmers.

"""
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50 , chunk_overlap = 20)
```

```python
chunk = splitter.split_text(text)
```

```python
from langchain_community.document_loaders import TextLoader , PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speech.txt", encoding="utf-8")
text = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/medical_report.pdf")
pdfs = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000 , chunk_overlap = 200)
```

```python
text_chunks = splitter.split_documents(text)
len(text_chunks)
```

```python
pdf_chunk = splitter.split_documents(pdfs)
len(pdf_chunk)
```
